## Dataset discovery

The following cell lists recognized tabular data files using paths relative to the notebook or project root. It does not load a file or infer a target.

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
RAW_DATA_DIR = PROJECT_ROOT / 'data' / 'raw'
SUPPORTED_SUFFIXES = {'.csv', '.tsv', '.xlsx', '.xls', '.json', '.parquet'}
data_files = sorted(
    path.relative_to(PROJECT_ROOT).as_posix()
    for path in RAW_DATA_DIR.rglob('*')
    if path.is_file() and path.suffix.lower() in SUPPORTED_SUFFIXES
) if RAW_DATA_DIR.exists() else []
print('Project root:', PROJECT_ROOT)
print('Supported data files:', data_files or 'None found')
```
